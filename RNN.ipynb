```python
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import SimpleRNN, LSTM, GRU, Dense
```

```python
# Generate sequence data
X = []
y = []
for i in range(100):
    seq = np.array([i, i+1, i+2,i+3])
    X.append(seq)
    y.append(i+3)

X = np.array(X)
y = np.array(y)

# Reshape for RNN input: [samples, timesteps, features]
X = X.reshape((X.shape[0], X.shape[1], 1))
```

```python
model_rnn = Sequential([
    SimpleRNN(10, activation='tanh', input_shape=(3, 1)),
    Dense(1)
])

model_rnn.compile(optimizer='adam', loss='mse')
model_rnn.fit(X, y, epochs=200, verbose=0)

print("RNN Prediction for [10, 11, 12]:", model_rnn.predict(np.array([[[10],[11],[12]]])))
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import SimpleRNN, Dense, Input
import numpy as np

# Define model
model_rnn = Sequential([
    Input(shape=(3, 1)),
    SimpleRNN(10, activation='tanh'),
    Dense(1)
])

# Compile and train
model_rnn.compile(optimizer='adam', loss='mse')
model_rnn.fit(X, y, epochs=200, verbose=0)

# Predict
print("RNN Prediction for [10, 11, 12]:",
      model_rnn.predict(np.array([[[10],[11],[12]]])))
```

```python
model_lstm = Sequential([
    LSTM(10, activation='tanh', input_shape=(3, 1)),
    Dense(1)
])

model_lstm.compile(optimizer='adam', loss='mse')
model_lstm.fit(X, y, epochs=200, verbose=0)

print("LSTM Prediction for [10, 11, 12]:", model_lstm.predict(np.array([[[10],[11],[12]]])))
```

```python
model_gru = Sequential([
    GRU(10, activation='tanh', input_shape=(3, 1)),
    Dense(1)
])

model_gru.compile(optimizer='adam', loss='mse')
model_gru.fit(X, y, epochs=200, verbose=0)

print("GRU Prediction for [10, 11, 12]:", model_gru.predict(np.array([[[10],[11],[12]]])))
```
